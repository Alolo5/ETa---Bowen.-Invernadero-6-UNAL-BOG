```python
import pandas as pd
```

```python
def procesar_con_pandas(archivo_entrada, archivo_salida):
    df = pd.read_csv(archivo_entrada)

    # Convertir a datetime y redondear hacia abajo a cada 10 min
    df["Time"] = pd.to_datetime(df["Time"]).dt.floor("10min")

    # Guardar el nuevo archivo respetando la estructura
    df.to_csv(archivo_salida, index=False)
```

```python
# Ejecutar para ambos archivos
procesar_con_pandas(
    "/home/alejandro/labtesis/DATA - acotada/datos_psicrometrico1_completos.txt",
    "/home/alejandro/labtesis/DATA - acotada/datos_psicrometrico1_redondeado.txt",
)
procesar_con_pandas(
    "/home/alejandro/labtesis/DATA - acotada/datos_psicrometrico2_completos.txt", 
    "/home/alejandro/labtesis/DATA - acotada/datos_psicrometrico2_redondeado.txt"
)
```
